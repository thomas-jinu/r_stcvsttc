```python
import csv
import math
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np


def read_td_csv(
    input_path: Path,
) -> dict[str, np.ndarray]:
    """
    Read the wide operator CSV.

    Each array has columns:
        [time, site_index, real_part, imaginary_part]
    """
    rows_by_operator: dict[str, list[list[float]]] = {}

    with input_path.open("r", newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        if reader.fieldnames is None:
            raise ValueError("CSV file has no header")

        labels = sorted(
            column.removesuffix("_real")
            for column in reader.fieldnames
            if column.endswith("_real")
        )

        for label in labels:
            rows_by_operator[label] = []

        for row in reader:
            time = float(row["t"])
            site = float(row["site_index"])

            for label in labels:
                real_text = row.get(f"{label}_real", "")
                imag_text = row.get(f"{label}_imag", "")

                if not real_text or not imag_text:
                    continue

                rows_by_operator[label].append(
                    [
                        time,
                        site,
                        float(real_text),
                        float(imag_text),
                    ]
                )

    return {
        label: np.asarray(rows, dtype=float)
        for label, rows in rows_by_operator.items()
    }


data = read_td_csv(
    Path("/Users/qqt/Library/CloudStorage/OneDrive-OakRidgeNationalLaboratory/Research_Projects/r_stcvsttc/code/template_folder_hubbard_chain_local_observables/N=4_Nup=2_Ndown=2_20260928_190559/td_N=4_Nup=2_Ndown=2_20260928_192851/runForinput_N=4_Nup=2_Ndown=2_20260928_192851_operators.csv"),
)

```
